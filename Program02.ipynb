```python
import numpy as np
import tensorflow as tf
```

```python
def print_immutable_tensor_details(tensor):
    print(f"tensor = {tensor}")
    print(f"size or tensor = {len(tensor)}")
    print(f"no of dimensions in tensor = {tensor.ndim}")
    print(f"shape of tensor = {tensor.shape}")
    print(f"data type of every element = {tensor.dtype}")
```

```python
def print_mutable_tensor_details(tensor):
    print(f"tensor = {tensor}")
    print(f"shape of tensor = {tensor.shape}")
    print(f"data type of every element = {tensor.dtype}")
```

```python
# create immutable tensor
tensor1 = tf.constant([10, 20, 30, 40])
print_immutable_tensor_details(tensor1)
```

```python
# create mutable tensor
tensor2 = tf.Variable([10, 20, 30, 40])
print_mutable_tensor_details(tensor2)
```
