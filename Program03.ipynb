```python
import numpy as numpy
import tensorflow as tf
```

```python
def print_tensor_details(tensor):
    print(f"tensor = {tensor}")
    print(f"no of dimensions in tensor = {tensor.ndim}")
    print(f"shape of tensor = {tensor.shape}")
    print(f"data type of every element = {tensor.dtype}")
    if tensor.ndim > 0:
        print(f"size or tensor = {len(tensor)}")
```

```python
num =tf.constant(50)
print_tensor_details(num)
```

```python
float_value =tf.constant(15.20)
print_tensor_details(float_value)
```

```python
# create a string scalar tensor
name = tf.constant("sunbeam")
print_tensor_details(name)
```

```python
# create a boolean scalar tensor
can_vote = tf.constant(True)
print_tensor_details(can_vote)
```

```python
numbers =tf.constant([10,20,30,40,50])
print_tensor_details(numbers)
```

```python
# create a vector with integer and float values
# implicit type conversion (int to float)
numbers = tf.constant([10, 20., 30, 40., 50])
print_tensor_details(numbers)
```

```python
countries = tf.constant(['india', 'us', 'uk'])
print_tensor_details(countries)
```

```python
matrix = tf.constant([
    [10,20],
    [30,40],
    [50,60]
])
print_tensor_details(matrix)
```
