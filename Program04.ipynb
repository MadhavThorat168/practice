```python
import numpy as np
import tensorflow as tf
```

```python
def print_tensor_details(tensor):
    print(f"tensor = {tensor}")
    print(f"size or tensor = {len(tensor)}")
    print(f"no of dimensions in tensor = {tensor.ndim}")
    print(f"shape of tensor = {tensor.shape}")
    print(f"data type of every element = {tensor.dtype}")
```

```python
vector=tf.zeros(5)
print_tensor_details(vector)
```

```python
matrix = tf.zeros([2, 3])
print_tensor_details(matrix)
```

```python
tensor_3d = tf.zeros([2, 3, 4])
print_tensor_details(tensor_3d)
```

```python
tensor_1d =tf.ones(5)
print_tensor_details(tensor_1d)
```

```python
tensor_2d =tf.ones([2,3])
print_tensor_details(tensor_2d)
```

```python
tensor_1d = tf.constant([10, 20, 30, 40, 50], dtype=tf.float16)
print_tensor_details(tensor_1d)
```

```python
tensor_1d = tf.zeros(5, dtype=tf.int16)
print_tensor_details(tensor_1d)
```

```python
tensor_1d = tf.ones(5, dtype=tf.int16)
print_tensor_details(tensor_1d)
```

```python
tensor_1d=tf.fill(5,7)
print_tensor_details(tensor_1d)
```

```python
tensor_2d=tf.fill((3,3),88)
print_tensor_details(tensor_2d)
```

```python
tensor=tf.constant([[10,20],[30,40],[50,60]])
```

```python
tensor_zeros=tf.zeros_like(tensor)
print_tensor_details(tensor_zeros)
```

```python
tensor_ones =tf.ones_like(tensor)
print_tensor_details(tensor_ones)
```

```python
tensor_sixes = tf.fill(tensor.shape,6)
print_tensor_details(tensor_sixes)
```
