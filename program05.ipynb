```python
import numpy as np
import tensorflow as tf
import pandas as pd
```

```python
def print_tensor_details(tensor):
    print(f"tensor={tensor}")
    print(f"size or tensor ={len(tensor)}")
    print (f" number of dimensions in tensor={tensor.ndim}")
    print(f" shape of tension {tensor.shape}") 
    print(f" data type of every element ={tensor.dtype}")
```

```python
array=np.array([10,20,30,40,50,60])
tensor_1d=tf.constant(array)
print_tensor_details(tensor_1d)
```

```python
# create a data frame (2d array) using pandas
df = pd.DataFrame({
    "feature1": [10, 20, 30, 40, 50],
    "feature2": [60, 70, 80, 90, 100]
})
df
```

```python
tensor_2d=tf.constant(df)
print_tensor_details(tensor_2d)
```
